Exercise 4.1 — Comparing Numbers

```python
secret = 7
guess = 5

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```

Exercise 4.2 — Small and Green

```python
small = True
green = True

if small:
    if green:
        print("pea")
    else:
        print("cherry")
else:
    if green:
        print("watermelon")
    else:
        print("pumpkin")
```

Exercise 6.1 — For Loop

Exercise 6.2 — While Loop

```python
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1
```

Exercise 6.3 — For Loop and Range

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
```
